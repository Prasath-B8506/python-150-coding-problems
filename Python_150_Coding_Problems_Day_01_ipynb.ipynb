# Q1  **is** “Is Even"


```python
num = int(input("Enter a number"))

if num %2 == 0:
  print("Is Even")
else:
  print("IS Odd")
```

# **Q4 - Return Second Last Digit**

```python
num = int(input("Enter a number"))

result = (num//10)%10
print(f"The secon last digit is {result}")
```

# **Q5** - Sum of last digit of two given numbers.

```python
num1 = int(input("Enter a number 1 "))
num2 = int(input("Enter a number 2 "))

result = (num1 % 10) + (num2 % 10)

print("Sum of last digit",result)
```

# **Q6 - Is N an Exact Multiple of M?**

```python
N = int(input("Enter a value of N : "))
M = int(input("Enter a value of M : "))

if N % M == 0:
  print ("Exact Multiple")
else:
  print ("It is not exact Multiple")
```
